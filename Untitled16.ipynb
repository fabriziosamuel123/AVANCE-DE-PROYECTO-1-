```python
# ============================================================
# SISTEMA DE INGRESO DE VARIABLES
# SEMÁFORO INTELIGENTE - ÓVALO HABICH
# ============================================================


# ============================================================
# FUNCIONES DE VALIDACIÓN
# ============================================================

def ingresar_entero(mensaje, minimo=0):
    while True:
        try:
            valor = int(input(mensaje))
            if valor < minimo:
                print(f"⚠️ El valor debe ser mayor o igual a {minimo}.")
            else:
                return valor
        except ValueError:
            print("⚠️ Ingresa un número entero válido.")


def ingresar_decimal(mensaje, minimo=0):
    while True:
        try:
            valor = float(input(mensaje))
            if valor < minimo:
                print(f"⚠️ El valor debe ser mayor o igual a {minimo}.")
            else:
                return valor
        except ValueError:
            print("⚠️ Ingresa un número válido.")


def elegir_opcion(mensaje, opciones):
    while True:

        print("\n" + mensaje)

        for numero, opcion in enumerate(opciones, start=1):
            print(f"{numero}. {opcion}")

        seleccion = ingresar_entero(
            "Seleccione una opción: ",
            1
        )

        if 1 <= seleccion <= len(opciones):
            return opciones[seleccion - 1]

        print("⚠️ Opción incorrecta.")


# ============================================================
# PUNTO 2
# VARIABLES DE DETECCIÓN VEHICULAR
# ============================================================

print("\n========================================")
print("PUNTO 2 - DETECCIÓN VEHICULAR")
print("========================================")

vehiculos_livianos = ingresar_entero(
    "Cantidad de vehículos livianos: "
)

motos = ingresar_entero(
    "Cantidad de motos: "
)

busetas = ingresar_entero(
    "Cantidad de busetas: "
)

buses = ingresar_entero(
    "Cantidad de buses: "
)

camiones_medianos = ingresar_entero(
    "Cantidad de camiones medianos: "
)

camiones_pesados = ingresar_entero(
    "Cantidad de camiones pesados: "
)

peatones_detectados = ingresar_entero(
    "Cantidad de peatones detectados: "
)


total_vehiculos = (
    vehiculos_livianos
    + motos
    + busetas
    + buses
    + camiones_medianos
    + camiones_pesados
)

print("\nTotal de vehículos detectados:", total_vehiculos)


# ============================================================
# PUNTO 3
# VARIABLES DE FLUJO VEHICULAR
# ============================================================

print("\n========================================")
print("PUNTO 3 - FLUJO VEHICULAR")
print("========================================")

flujo_norte_sur = ingresar_entero(
    "Flujo Norte-Sur: "
)

flujo_sur_norte = ingresar_entero(
    "Flujo Sur-Norte: "
)

flujo_este_oeste = ingresar_entero(
    "Flujo Este-Oeste: "
)

flujo_oeste_este = ingresar_entero(
    "Flujo Oeste-Este: "
)


# ============================================================
# PUNTO 4
# VARIABLES DE VELOCIDAD
# ============================================================

print("\n========================================")
print("PUNTO 4 - VELOCIDAD PROMEDIO")
print("========================================")

velocidad_promedio = ingresar_decimal(
    "Ingrese la velocidad promedio en km/h: "
)

VELOCIDAD_EXTREMA_MAX = 10
VELOCIDAD_ALTA_MAX = 20
VELOCIDAD_MEDIA_MAX = 30


# ============================================================
# PUNTO 5
# NIVEL DE CONGESTIÓN
# ============================================================

print("\n========================================")
print("PUNTO 5 - NIVEL DE CONGESTIÓN")
print("========================================")


# Se puede determinar automáticamente
# según la velocidad ingresada.

if velocidad_promedio <= VELOCIDAD_EXTREMA_MAX:

    nivel_congestion = "EXTREMA"

elif velocidad_promedio <= VELOCIDAD_ALTA_MAX:

    nivel_congestion = "ALTA"

elif velocidad_promedio <= VELOCIDAD_MEDIA_MAX:

    nivel_congestion = "MEDIA"

else:

    nivel_congestion = "BAJA"


print(
    "Nivel de congestión calculado:",
    nivel_congestion
)


# ============================================================
# PUNTO 6
# UMBRALES DE DEMANDA VEHICULAR
# ============================================================

print("\n========================================")
print("PUNTO 6 - UMBRALES DE DEMANDA")
print("========================================")

print("""
Valores de referencia:

Muy bajo : hasta 5 vehículos/min
Bajo     : 6 a 17 vehículos/min
Medio    : 18 a 29 vehículos/min
Alto     : 30 a 42 vehículos/min
Muy alto : 43 o más vehículos/min
""")


usar_umbrales = elegir_opcion(
    "¿Qué deseas hacer?",
    [
        "Usar valores establecidos",
        "Ingresar valores personalizados"
    ]
)


if usar_umbrales == "Usar valores establecidos":

    UMBRAL_MUY_BAJO = 6
    UMBRAL_BAJO = 18
    UMBRAL_MEDIO = 30
    UMBRAL_ALTO = 43

else:

    UMBRAL_MUY_BAJO = ingresar_entero(
        "Inicio de demanda baja: "
    )

    UMBRAL_BAJO = ingresar_entero(
        "Inicio de demanda media: "
    )

    UMBRAL_MEDIO = ingresar_entero(
        "Inicio de demanda alta: "
    )

    UMBRAL_ALTO = ingresar_entero(
        "Inicio de demanda muy alta: "
    )


# ============================================================
# PUNTO 7
# TIEMPOS DE LUZ VERDE
# ============================================================

print("\n========================================")
print("PUNTO 7 - TIEMPOS DE LUZ VERDE")
print("========================================")

print("""
Valores iniciales de referencia:

Muy baja demanda = 5 segundos
Baja demanda     = 15 segundos
Media demanda    = 30 segundos
Alta demanda     = 45 segundos
Muy alta demanda = 60 segundos
""")


opcion_tiempos = elegir_opcion(
    "Seleccione:",
    [
        "Usar tiempos establecidos",
        "Ingresar tiempos personalizados"
    ]
)


if opcion_tiempos == "Usar tiempos establecidos":

    VERDE_MUY_BAJO = 5
    VERDE_BAJO = 15
    VERDE_MEDIO = 30
    VERDE_ALTO = 45
    VERDE_MUY_ALTO = 60

else:

    VERDE_MUY_BAJO = ingresar_entero(
        "Tiempo verde para demanda muy baja: "
    )

    VERDE_BAJO = ingresar_entero(
        "Tiempo verde para demanda baja: "
    )

    VERDE_MEDIO = ingresar_entero(
        "Tiempo verde para demanda media: "
    )

    VERDE_ALTO = ingresar_entero(
        "Tiempo verde para demanda alta: "
    )

    VERDE_MUY_ALTO = ingresar_entero(
        "Tiempo verde para demanda muy alta: "
    )


# ============================================================
# CLASIFICACIÓN AUTOMÁTICA DE LA DEMANDA
# ============================================================

# Se usa el total de vehículos introducido en el punto 2.

if total_vehiculos < UMBRAL_MUY_BAJO:

    demanda = "MUY BAJA"
    tiempo_verde_asignado = VERDE_MUY_BAJO

elif total_vehiculos < UMBRAL_BAJO:

    demanda = "BAJA"
    tiempo_verde_asignado = VERDE_BAJO

elif total_vehiculos < UMBRAL_MEDIO:

    demanda = "MEDIA"
    tiempo_verde_asignado = VERDE_MEDIO

elif total_vehiculos < UMBRAL_ALTO:

    demanda = "ALTA"
    tiempo_verde_asignado = VERDE_ALTO

else:

    demanda = "MUY ALTA"
    tiempo_verde_asignado = VERDE_MUY_ALTO


print("\nDemanda calculada:", demanda)
print(
    "Tiempo verde recomendado:",
    tiempo_verde_asignado,
    "segundos"
)


# ============================================================
# PUNTO 8
# VARIABLES DEL SEMÁFORO
# ============================================================

print("\n========================================")
print("PUNTO 8 - VARIABLES DEL SEMÁFORO")
print("========================================")


fase_vehicular = elegir_opcion(
    "Seleccione la fase vehicular:",
    [
        "ROJO",
        "AMARILLO",
        "VERDE"
    ]
)


fase_peatonal = elegir_opcion(
    "Seleccione la fase peatonal:",
    [
        "ROJO",
        "VERDE"
    ]
)


tiempo_verde_asignado = ingresar_entero(
    f"Ingrese tiempo verde asignado "
    f"[recomendado {tiempo_verde_asignado}s]: "
)


tiempo_restante = ingresar_entero(
    "Ingrese tiempo restante de la fase: "
)


via_prioritaria = elegir_opcion(
    "Seleccione la vía prioritaria:",
    [
        "NORTE-SUR",
        "SUR-NORTE",
        "ESTE-OESTE",
        "OESTE-ESTE"
    ]
)


# ============================================================
# PUNTO 9
# ESTADOS DEL SEMÁFORO
# ============================================================

print("\n========================================")
print("PUNTO 9 - ESTADO DEL SEMÁFORO")
print("========================================")


estado_semaforo = elegir_opcion(
    "Seleccione el estado actual:",
    [
        "ROJO",
        "AMARILLO",
        "VERDE"
    ]
)


# ============================================================
# PUNTO 10
# VEHÍCULOS POR ACCESO
# ============================================================

print("\n========================================")
print("PUNTO 10 - VEHÍCULOS POR ACCESO")
print("========================================")


vehiculos_norte_sur = ingresar_entero(
    "Vehículos Norte-Sur: "
)

vehiculos_sur_norte = ingresar_entero(
    "Vehículos Sur-Norte: "
)

vehiculos_este_oeste = ingresar_entero(
    "Vehículos Este-Oeste: "
)

vehiculos_oeste_este = ingresar_entero(
    "Vehículos Oeste-Este: "
)


# Determinar automáticamente vía con mayor cantidad
vias = {

    "NORTE-SUR":
        vehiculos_norte_sur,

    "SUR-NORTE":
        vehiculos_sur_norte,

    "ESTE-OESTE":
        vehiculos_este_oeste,

    "OESTE-ESTE":
        vehiculos_oeste_este
}


via_mayor_demanda = max(
    vias,
    key=vias.get
)


print(
    "\nVía con mayor demanda:",
    via_mayor_demanda
)

print(
    "Vehículos:",
    vias[via_mayor_demanda]
)


# ============================================================
# PUNTO 11
# TIEMPOS DE ESPERA
# ============================================================

print("\n========================================")
print("PUNTO 11 - TIEMPOS DE ESPERA")
print("========================================")


tiempo_espera_norte_sur = ingresar_decimal(
    "Tiempo de espera Norte-Sur (segundos): "
)

tiempo_espera_sur_norte = ingresar_decimal(
    "Tiempo de espera Sur-Norte (segundos): "
)

tiempo_espera_este_oeste = ingresar_decimal(
    "Tiempo de espera Este-Oeste (segundos): "
)

tiempo_espera_oeste_este = ingresar_decimal(
    "Tiempo de espera Oeste-Este (segundos): "
)


# ============================================================
# PUNTO 12
# COLA VEHICULAR
# ============================================================

print("\n========================================")
print("PUNTO 12 - COLA VEHICULAR")
print("========================================")


cola_norte_sur = ingresar_entero(
    "Vehículos en cola Norte-Sur: "
)

cola_sur_norte = ingresar_entero(
    "Vehículos en cola Sur-Norte: "
)

cola_este_oeste = ingresar_entero(
    "Vehículos en cola Este-Oeste: "
)

cola_oeste_este = ingresar_entero(
    "Vehículos en cola Oeste-Este: "
)


# ============================================================
# PUNTO 14
# PARÁMETROS DE CALIBRACIÓN
# ============================================================

print("\n========================================")
print("PUNTO 14 - CALIBRACIÓN DEL SISTEMA")
print("========================================")


CONFIANZA_MINIMA = ingresar_decimal(
    "Confianza mínima de YOLO (ejemplo 0.50): "
)


while CONFIANZA_MINIMA > 1:

    print(
        "⚠️ La confianza debe estar "
        "entre 0 y 1."
    )

    CONFIANZA_MINIMA = ingresar_decimal(
        "Confianza mínima: "
    )


INTERVALO_EVALUACION = ingresar_decimal(
    "Intervalo de evaluación en segundos: "
)


# ============================================================
# RESUMEN FINAL
# ============================================================

print("\n")
print("=" * 60)
print("RESUMEN DE VARIABLES INGRESADAS")
print("=" * 60)


print("\n--- PUNTO 2: DETECCIÓN ---")

print("Vehículos livianos:", vehiculos_livianos)
print("Motos:", motos)
print("Busetas:", busetas)
print("Buses:", buses)
print("Camiones medianos:", camiones_medianos)
print("Camiones pesados:", camiones_pesados)
print("Peatones:", peatones_detectados)
print("TOTAL VEHÍCULOS:", total_vehiculos)


print("\n--- PUNTO 3: FLUJO ---")

print("Norte-Sur:", flujo_norte_sur)
print("Sur-Norte:", flujo_sur_norte)
print("Este-Oeste:", flujo_este_oeste)
print("Oeste-Este:", flujo_oeste_este)


print("\n--- PUNTO 4 Y 5 ---")

print(
    "Velocidad promedio:",
    velocidad_promedio,
    "km/h"
)

print(
    "Nivel de congestión:",
    nivel_congestion
)


print("\n--- PUNTOS 6 Y 7 ---")

print(
    "Nivel de demanda:",
    demanda
)

print(
    "Tiempo verde:",
    tiempo_verde_asignado,
    "segundos"
)


print("\n--- PUNTOS 8 Y 9 ---")

print(
    "Fase vehicular:",
    fase_vehicular
)

print(
    "Fase peatonal:",
    fase_peatonal
)

print(
    "Estado actual:",
    estado_semaforo
)

print(
    "Vía prioritaria:",
    via_prioritaria
)


print("\n--- PUNTO 10 ---")

print("Norte-Sur:", vehiculos_norte_sur)
print("Sur-Norte:", vehiculos_sur_norte)
print("Este-Oeste:", vehiculos_este_oeste)
print("Oeste-Este:", vehiculos_oeste_este)

print(
    "Vía con mayor cantidad:",
    via_mayor_demanda
)


print("\n--- PUNTO 11 ---")

print(
    "Espera Norte-Sur:",
    tiempo_espera_norte_sur,
    "s"
)

print(
    "Espera Sur-Norte:",
    tiempo_espera_sur_norte,
    "s"
)

print(
    "Espera Este-Oeste:",
    tiempo_espera_este_oeste,
    "s"
)

print(
    "Espera Oeste-Este:",
    tiempo_espera_oeste_este,
    "s"
)


print("\n--- PUNTO 12 ---")

print(
    "Cola Norte-Sur:",
    cola_norte_sur
)

print(
    "Cola Sur-Norte:",
    cola_sur_norte
)

print(
    "Cola Este-Oeste:",
    cola_este_oeste
)

print(
    "Cola Oeste-Este:",
    cola_oeste_este
)


print("\n--- PUNTO 14 ---")

print(
    "Confianza mínima:",
    CONFIANZA_MINIMA
)

print(
    "Intervalo de evaluación:",
    INTERVALO_EVALUACION,
    "segundos"
)


print("\n" + "=" * 60)
print("✅ INGRESO DE DATOS COMPLETADO")
print("=" * 60)
```
